```python
import pandas as pd

clientes = pd.read_csv("clientes.csv")
productos = pd.read_csv("productos_.csv")
ventas = pd.read_csv("ventas.csv")
```

1. **Líneas, tickets y unidades.** ¿Cuántas líneas de venta, cuántos tickets distintos y cuántas unidades vendidas hay en el archivo? Explique qué representa cada conteo.

```python
lventa = len(ventas)
tdistinto = len (ventas["id_ticket"].unique())
cantidadesv = ventas["cantidad"].sum()

print ("El numero de Las lineas de venta son: ", lventa)
print ("El numero de Los tickets distintos son: ", tdistinto)
print ("El numero de Las unidades vendidas son: ", cantidadesv)
```

2. **Tickets por sucursal.** ¿Cuántos tickets se registraron en cada sucursal? ¿Cuál registró más?

```python
txsucursal = ventas.groupby("sucursal").count()["id_ticket"]

print ("numero de tickets de la sucursal Central: ", txsucursal.Central)
print ("numero de tickets de la sucursal Norte: ", txsucursal.Norte)
print ("numero de tickets de la sucursal Sur: ", txsucursal.Sur)

maxsucursal = txsucursal.idxmax()
print ("La central q mas registro fue ",maxsucursal)
```

4. **Precios de lista por categoría.** ¿Cuál es el precio de lista mínimo, máximo y promedio de los productos de cada categoría? Use el catálogo de productos.

```python
pmin = productos["precio_lista"].min()
pmax = productos["precio_lista"].max()
pprom = productos["precio_lista"].mean()

print("El precio minimo es de: ",pmin)
print("El precio maximo es de: ",pmax)
print("El precio promedio es de: ",pprom)
```

6. **Importe de ventas por sucursal.** ¿Cuál fue el importe total de ventas de cada sucursal? Calcule primero el precio total de cada línea como `cantidad × precio_unitario`.

```python
ventas ["totalGanancia"] = ventas["precio_unitario"] * ventas["cantidad"]
#print (ventas)
preciosSucursal = ventas.groupby("sucursal")["totalGanancia"].sum()

print ("total ventas Central: ", preciosSucursal.Central)
print ("total ventas Norte: ", preciosSucursal.Norte)
print ("total ventas Sur: ", preciosSucursal.Sur)
```
